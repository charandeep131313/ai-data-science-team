```python
import pandas as pd
import yfinance as yf
import plotly.graph_objects as go
import numpy as np
```

```python
btc_data = yf.download(tickers='BTC-USD', period='5d', interval='5m')
```

```python
btc_data['MA20'] = btc_data['Close'].rolling(window=20).mean()
btc_data['MA50'] = btc_data['Close'].rolling(window=50).mean()
```

```python
btc_data['signal'] = 0.0
btc_data['signal'] = np.where(btc_data['MA20'] > btc_data['MA50'], 1.0, 0.0)
btc_data['positions'] = btc_data['signal'].diff()
btc_data['buy_signal_price'] = btc_data['Close'].where(btc_data['positions'] == 1)
btc_data['sell_signal_price'] = btc_data['Close'].where(btc_data['positions'] == -1)
```

```python
fig = go.Figure(data=[
    go.Scatter(x=btc_data.index, y=btc_data['Close'], name='BTC Price'),
    go.Scatter(x=btc_data.index, y=btc_data['MA20'], name='MA20'),
    go.Scatter(x=btc_data.index, y=btc_data['MA50'], name='MA50'),
    go.Scatter(x=btc_data.index, y=btc_data['buy_signal_price'], name='Buy Signal', mode='markers', marker=dict(symbol='triangle-up', color='green', size=10)),
    go.Scatter(x=btc_data.index, y=btc_data['sell_signal_price'], name='Sell Signal', mode='markers', marker=dict(symbol='triangle-down', color='red', size=10))
])
fig.update_layout(title='BTC-USD Intraday Strategy (5m)', yaxis_title='Price (USD)', xaxis_title='Date')
fig.show()
```
