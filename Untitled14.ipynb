Provide the complete trading bot script, including the main loop, `trade_logic` and `in_session` functions, and execute it to fetch data, apply indicators, check for signals, and simulate trades using the provided configuration for `PAPER_TRADE` and `CAPITAL`.

````python
import ccxt
import pandas as pd
import pandas_ta as ta
import time
from datetime import datetime
import logging
from typing import Tuple, Optional

# ======================
# LOGGING SETUP
# ======================
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    handlers=[
        logging.FileHandler('trading_bot.log'),
        logging.StreamHandler()
    ]
)
logger = logging.getLogger(__name__)

# ======================
# CONFIG
# ======================
SYMBOL = "BTC/USDT"
TIMEFRAME = "5m"
LOOKBACK = 15
RSI_LEN = 14
ATR_LEN = 14
ATR_MULT = 0.8
RR = 1.8
RISK_PER_TRADE = 0.01   # 1% of available balance

# Strategy filters
RSI_LONG_MIN, RSI_LONG_MAX = 50, 55
RSI_SHORT_MIN, RSI_SHORT_MAX = 45, 50

# Safety limits
MIN_POSITION_SIZE = 0.001  # Minimum BTC to trade
MAX_POSITION_SIZE = 0.5    # Maximum BTC to trade
COOLDOWN_PERIOD = 300      # 5 min between signals

# ======================
# EXCHANGE (READ-ONLY)
# ======================
exchange = ccxt.mexc({
    "enableRateLimit": True,
    "options": {
        "defaultType": "spot",  # Change to "swap" for futures
        "adjustForTimeDifference": True  # Handle time sync issues
    }
})

# ======================
# STATE MANAGEMENT
# ======================
class BotState:
    def __init__(self):
        self.last_signal_time = 0
        self.trades_today = 0
        self.consecutive_errors = 0

    def can_trade(self) -> bool:
        """Check if enough time has passed since last signal"""
        return (time.time() - self.last_signal_time) > COOLDOWN_PERIOD

    def record_signal(self):
        """Record that a signal was generated"""
        self.last_signal_time = time.time()
        self.trades_today += 1

state = BotState()

# ======================
# FETCH OHLCV WITH ERROR HANDLING
# ======================
def fetch_data() -> Optional[pd.DataFrame]:
    """Fetch OHLCV data with proper error handling"""
    try:
        ohlcv = exchange.fetch_ohlcv(SYMBOL, TIMEFRAME, limit=100)
        if not ohlcv or len(ohlcv) < max(LOOKBACK, RSI_LEN, ATR_LEN) + 1:
            logger.warning("Insufficient data received")
            return None

        df = pd.DataFrame(ohlcv, columns=[
            "time", "open", "high", "low", "close", "volume"
        ])
        df['time'] = pd.to_datetime(df['time'], unit='ms')
        return df
    except ccxt.NetworkError as e:
        logger.error(f"Network error fetching data: {e}")
        return None
    except Exception as e:
        logger.error(f"Error fetching data: {e}")
        return None

# ======================
# FETCH AVAILABLE BALANCE
# ======================
def fetch_balance() -> Optional[float]:
    """Fetch available USDT balance with error handling"""
    try:
        balance = exchange.fetch_balance()
        usdt_balance = balance["free"]["USDT"]
        if usdt_balance <= 0:
            logger.warning("Zero or negative balance detected")
            return None
        return usdt_balance
    except Exception as e:
        logger.error(f"Error fetching balance: {e}")
        return None

# ======================
# ENHANCED STRATEGY LOGIC
# ======================
def calculate_indicators(df: pd.DataFrame) -> pd.DataFrame:
    """Calculate all technical indicators"""
    df = df.copy()

    # Set datetime index for VWAP calculation
    if not isinstance(df.index, pd.DatetimeIndex):
        df.set_index('time', inplace=True)

    # Core indicators
    df["vwap"] = ta.vwap(df["high"], df["low"], df["close"], df["volume"])
    df["rsi"] = ta.rsi(df["close"], RSI_LEN)
    df["atr"] = ta.atr(df["high"], df["low"], df["close"], ATR_LEN)

    # Key levels
    df["key_high"] = df["high"].rolling(LOOKBACK).max().shift(1)
    df["key_low"] = df["low"].rolling(LOOKBACK).min().shift(1)

    # Additional filters (volume confirmation)
    df["volume_ma"] = df["volume"].rolling(20).mean()
    df["volume_ratio"] = df["volume"] / df["volume_ma"]

    return df

def check_signal(df: pd.DataFrame) -> Tuple[Optional[str], Optional[pd.Series]]:
    """Check for trading signals with enhanced filters"""
    if df is None or len(df) < 2:
        return None, None

    df = calculate_indicators(df)

    # Remove NaN rows
    df = df.dropna()
    if len(df) < 2:
        return None, None

    last = df.iloc[-1]
    prev = df.iloc[-2]

    # Check for NaN in critical values
    critical_values = [last["close"], last["vwap"], last["rsi"], last["atr"]]
    if any(pd.isna(val) for val in critical_values):
        logger.warning("NaN values detected in indicators")
        return None, None

    # Volume filter (at least 80% of average)
    if last["volume_ratio"] < 0.8:
        return None, None

    # LONG CONDITIONS
    long_conditions = [
        last["close"] > last["vwap"],
        last["close"] > last["key_high"],
        RSI_LONG_MIN < last["rsi"] < RSI_LONG_MAX,
        last["close"] > prev["close"],  # Momentum confirmation
    ]

    if all(long_conditions):
        return "LONG", last

    # SHORT CONDITIONS
    short_conditions = [
        last["close"] < last["vwap"],
        last["close"] < last["key_low"],
        RSI_SHORT_MIN < last["rsi"] < RSI_SHORT_MAX,
        last["close"] < prev["close"],  # Momentum confirmation
    ]

    if all(short_conditions):
        return "SHORT", last

    return None, None

# ======================
# POSITION SIZE WITH SAFETY CHECKS
# ======================
def calc_position_size(entry: float, sl: float, balance: float) -> Optional[float]:
    """Calculate position size with safety limits"""
    try:
        risk_amount = balance * RISK_PER_TRADE
        risk_per_unit = abs(entry - sl)

        if risk_per_unit == 0:
            logger.error("Stop loss equals entry price")
            return None

        size = risk_amount / risk_per_unit

        # Apply size limits
        if size < MIN_POSITION_SIZE:
            logger.warning(f"Position size {size} below minimum {MIN_POSITION_SIZE}")
            return None

        if size > MAX_POSITION_SIZE:
            logger.warning(f"Position size {size} capped at maximum {MAX_POSITION_SIZE}")
            size = MAX_POSITION_SIZE

        # Round to appropriate precision for the exchange
        return round(size, 4)

    except Exception as e:
        logger.error(f"Error calculating position size: {e}")
        return None

# ======================
# RISK VALIDATION
# ======================
def validate_trade(signal: str, entry: float, sl: float, tp: float, size: float) -> bool:
    """Validate trade parameters before execution"""

    # Check SL distance (not too tight, not too wide)
    sl_distance_pct = abs(entry - sl) / entry * 100
    if sl_distance_pct < 0.1 or sl_distance_pct > 5.0:
        logger.warning(f"SL distance {sl_distance_pct:.2f}% outside acceptable range")
        return False

    # Check TP is on correct side
    if signal == "LONG" and tp <= entry:
        logger.error("TP below entry for LONG")
        return False
    if signal == "SHORT" and tp >= entry:
        logger.error("TP above entry for SHORT")
        return False

    # Check SL is on correct side
    if signal == "LONG" and sl >= entry:
        logger.error("SL above entry for LONG")
        return False
    if signal == "SHORT" and sl <= entry:
        logger.error("SL below entry for SHORT")
        return False

    return True

# ======================
# MAIN LOOP
# ======================
def main():
    """Main trading loop"""
    logger.info("🚀 VWAP Scalping Bot Started (AUTO BALANCE MODE)")
    logger.info(f"Symbol: {SYMBOL} | Timeframe: {TIMEFRAME}")
    logger.info(f"Risk per trade: {RISK_PER_TRADE*100}% | R:R = {RR}")

    while True:
        try:
            # Reset error counter on successful iteration
            if state.consecutive_errors > 0:
                state.consecutive_errors = 0

            # Fetch market data
            df = fetch_data()
            if df is None:
                time.sleep(60)
                continue

            # Check for signals
            signal, candle = check_signal(df)

            if signal and state.can_trade():
                # Fetch balance
                balance = fetch_balance()
                if balance is None or balance < 10:  # Minimum $10
                    logger.warning("Insufficient balance to trade")
                    time.sleep(60)
                    continue

                # Calculate trade parameters
                entry = float(candle["close"])
                atr = float(candle["atr"])

                if signal == "LONG":
                    sl = entry - atr * ATR_MULT
                    tp = entry + (entry - sl) * RR
                else:
                    sl = entry + atr * ATR_MULT
                    tp = entry - (sl - entry) * RR

                # Calculate position size
                size = calc_position_size(entry, sl, balance)
                if size is None:
                    time.sleep(60)
                    continue

                # Validate trade
                if not validate_trade(signal, entry, sl, tp, size):
                    time.sleep(60)
                    continue

                # Log trade details
                logger.info("\n" + "="*50)
                logger.info("📊 TRADE SIGNAL GENERATED")
                logger.info(f"Time    : {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
                logger.info(f"Balance : ${balance:.2f}")
                logger.info(f"Type    : {signal}")
                logger.info(f"Entry   : ${entry:.2f}")
                logger.info(f"SL      : ${sl:.2f} ({abs(entry-sl)/entry*100:.2f}%)")
                logger.info(f"TP      : ${tp:.2f} ({abs(tp-entry)/entry*100:.2f}%)")
                logger.info(f"Size    : {size} BTC (${size*entry:.2f})")
                logger.info(f"Risk    : ${size*abs(entry-sl):.2f}")
                logger.info(f"RSI     : {candle['rsi']:.2f}")
                logger.info(f"ATR     : ${atr:.2f}")
                logger.info("="*50 + "\n")

                # Record signal
                state.record_signal()

            # Sleep until next check
            time.sleep(60)

        except KeyboardInterrupt:
            logger.info("Bot stopped by user")
            break

        except Exception as e:
            state.consecutive_errors += 1
            logger.error(f"Unexpected error: {e}", exc_info=True)

            # If too many consecutive errors, increase sleep time
            if state.consecutive_errors > 5:
                logger.warning("Multiple consecutive errors, sleeping for 5 minutes")
                time.sleep(300)
            else:
                time.sleep(60)

if __name__ == "__main__":
    main()
````
